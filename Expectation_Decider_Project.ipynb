## 1. Understanding the Basics

**What is Probability?**
 
 Probability is a statistical measure that represents the chance of an event occurring. Its value lies between 0 (impossible) and 1 (certain).

**Terminology:**
- **Experiment:** A process whose result is uncertain (e.g., taking an exam).
- **Outcome:** An outcome of an experiment (e.g., passing).
- **Sample Space:** The set of all possible outcomes {Pass, Fail}.

**Examples from Dataset:**
1. A student's “passing.”
2. A student getting a “Yes” in the group discussion.
3. The previous test score being more than 80.

---
## 2. Types of Events

- **Theoretical Probability:** If we assume that the results are perfectly random, then P(Pass) = 1/2 = 0.5.
- **Empirical Probability:** According to the dataset, P(Pass) = 100/200 = 0.50. 

---
## 3. Random Variable & Probability Distribution

Let **X** = Number of students passing out of 3 selected students.

**Distribution Table:**
| X | P(X) |
|---|------|
| 0 | 0.1250 |
| 1 | 0.3750 |
| 2 | 0.3750 |
| 3 | 0.1250 |

**Mean (E[X]):** 1.50
**Variance:** 0.7500

---
## 4. Venn Diagram Logic

- Students (Study > 10hrs): 139
- Students (Attendance > 80%): 113
- Overlap (Both): 82

---
## 5. Contingency Table

| Group Discussion | Pass | Fail | Total |
|---|---|---|---|
| Yes | 58 | 25 | 83 |
| No | 42 | 75 | 117 |

- **Joint Prob (Yes & Pass):** 0.2900
- **Marginal Prob (Pass):** 0.50
- **Conditional Prob (Pass | Yes):** 0.6988

---
## 6. Relationships
Since P(Pass|Yes) and P(Pass) are not equal, these events are Dependent.

---
## 7. Bayes Theorem
P(Pass | High Attendance) = (P(High Attendance | Pass) * P(Pass)) / P(High Attendance)
Result: (0.70 * 0.50) / 0.60 = 0.5833

---
## Final Summary
The analysis shows that regular attendance and participation in group discussions significantly increase the probability of students’ success.
